## Ejercicio 1 · Calcular subtotal

Implementa `calcular_subtotal(precio_unitario, cantidad)` que devuelva el producto de precio por cantidad.

Si `precio_unitario <= 0` o `cantidad <= 0`, devuelve `None`.

**Ejemplo:** `calcular_subtotal(25000, 4)` → `100000`;
`calcular_subtotal(0, 5)` → `None`.

````python
FUNCIÓN calcular_subtotal(precio_unitario, cantidad):

    SI precio_unitario <= 0 O cantidad <= 0 ENTONCES
        total = NULO
    SINO
        total = precio_unitario * cantidad
    DEVOLVER total
````

````python
def calcular_subtotal(precio_unitario, cantidad):
    if precio_unitario <= 0 or cantidad <= 0:
      subtotal = None
    else:
      subtotal = precio_unitario * cantidad
    return subtotal
````

````python
# === Tests visibles · Ejercicio 1 ===
assert calcular_subtotal(25000, 4) == 100000
assert calcular_subtotal(0, 5) is None
print("✅ Ejercicio 1: tests visibles superados.")
````

## Ejercicio 2 · Aplicar descuento por método de pago

Implementa `descuento_metodo(subtotal, metodo_pago)` que devuelva el porcentaje de descuento según el método:
- `'efectivo'` → 5% (0.05) de descuento
- `'tarjeta'` → 2% (0.02) de descuento
- `'transferencia'` → 3% (0.03) de descuento
- cualquier otro → 0% de descuento

La función devuelve el **monto del descuento** (no el porcentaje).
Redondea a 2 decimales.

**Ejemplo:** `descuento_metodo(100000, 'efectivo')` → `5000.0`;
`descuento_metodo(80000, 'tarjeta')` → `1600.0`.

````python
FUNCIÓN descuento_metodo(subtotal, metodo_pago):

    SI metodo_pago == "efectivo" ENTONCES
        descuento = 0.05
    SINO SI metodo_pago == "tarjeta" ENTONCES
        descuento = 0.02
    SINO SI metodo_pago == "transferencia" ENTONCES
        descuento = 0.03
    SINO
        descuento = 0

    DEVOLVER subtotal * descuento
````

````python
def descuento_metodo(subtotal, metodo_pago):
    if metodo_pago == "efectivo":
      descuento = 0.05
    elif metodo_pago == "tarjeta":
      descuento = 0.02
    elif metodo_pago == "transferencia":
      descuento = 0.03
    else:
      descuento = 0
    return descuento *subtotal
````

````python
# === Tests visibles · Ejercicio 2 ===
assert descuento_metodo(100000, 'efectivo') == 5000.0
assert descuento_metodo(80000, 'tarjeta') == 1600.0
print("✅ Ejercicio 2: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 2 ===
assert descuento_metodo(100000, 'transferencia') == 3000.0
assert descuento_metodo(200000, 'criptomoneda') == 0.0
assert descuento_metodo(0, 'efectivo') == 0.0
print("✅ Ejercicio 2: tests adicionales superados.")
````

## Ejercicio 3 · Calcular IVA sobre base gravable

Implementa `calcular_iva_base(base_gravable, tasa_iva=0.19)` que devuelva el monto del IVA sobre la `base_gravable`.
Redondea a 2 decimales.

Si `base_gravable < 0` o `tasa_iva < 0`, devuelve `None`.

**Ejemplo:** `calcular_iva_base(95000)` → `18050.0`;
`calcular_iva_base(50000, 0.08)` → `4000.0`.

````python
FUNCIÓN calcular_iva_base(base_gravable, tasa_iva = 0.19):

    SI base_gravable < 0 O tasa_iva < 0 ENTONCES
        DEVOLVER NULO
    SINO
        DEVOLVER base_gravable * tasa_iva
````

````python
def calcular_iva_base(base_gravable, tasa_iva=0.19):
    if base_gravable < 0 or tasa_iva < 0:
      calcular_iva_base = None
    else:
      return base_gravable * tasa_iva
````

## Ejercicio 4 · Total de la compra

Implementa `total_compra(precio_unitario, cantidad, metodo_pago, tasa_iva=0.19)` que devuelva una tupla `(subtotal, descuento, base_gravable, iva, total)` usando las funciones anteriores:

1. `subtotal = calcular_subtotal(precio_unitario, cantidad)`
2. `descuento = descuento_metodo(subtotal, metodo_pago)`
3. `base_gravable = subtotal - descuento`
4. `iva = calcular_iva_base(base_gravable, tasa_iva)`
5. `total = base_gravable + iva`

Si `calcular_subtotal` devuelve `None`, toda la tupla es `None`.

**Ejemplo:** `total_compra(25000, 4, 'efectivo')` →
`(100000, 5000.0, 95000.0, 18050.0, 113050.0)`

````python
FUNCIÓN total_compra(precio_unitario, cantidad, metodo_pago, tasa_iva = 0.19):

    subtotal = calcular_subtotal(precio_unitario, cantidad)

    SI subtotal ES NULO ENTONCES
        DEVOLVER NULO

    descuento = descuento_metodo(subtotal, metodo_pago)
    base_gravable = subtotal - descuento
    iva = calcular_iva_base(base_gravable, tasa_iva)
    total = base_gravable + iva

    DEVOLVER tupla (subtotal, descuento, base_gravable, iva, total)
````

````python
def total_compra(precio_unitario, cantidad, metodo_pago, tasa_iva=0.19):
    subtotal = calcular_subtotal(precio_unitario, cantidad)
    if subtotal is None:
        return None
    else:
     descuento = descuento_metodo(subtotal, metodo_pago)
     base_gravable = subtotal - descuento
     iva = calcular_iva_base(base_gravable, tasa_iva)
     total = base_gravable + iva
    return (subtotal, descuento, base_gravable, iva, total)

````

````python
# === Tests visibles · Ejercicio 4 ===
_r = total_compra(25000, 4, 'efectivo')
assert _r == (100000, 5000.0, 95000.0, 18050.0, 113050.0)
print("✅ Ejercicio 4: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 4 ===
_r2 = total_compra(50000, 2, 'tarjeta')
assert _r2[0] == 100000
assert _r2[1] == 2000.0
assert _r2[4] == 116620.0
assert total_compra(0, 5, 'efectivo') is None
print("✅ Ejercicio 4: tests adicionales superados.")
````

## Ejercicio 5 · Calcular cambio

Implementa `calcular_cambio(total_a_pagar, monto_pagado)` que devuelva el **cambio** que se le debe al cliente.

- Si `monto_pagado >= total_a_pagar`, devuelve `monto_pagado - total_a_pagar` (redondeado a 2 decimales).
- Si `monto_pagado < total_a_pagar`, devuelve `None` (pago insuficiente).

**Ejemplos:** `calcular_cambio(113050.0, 150000)` → `36950.0`;
`calcular_cambio(113050.0, 100000)` → `None` (no alcanza).

````python
FUNCIÓN calcular_cambio(total_a_pagar, monto_pagado):

    SI monto_pagado >= total_a_pagar ENTONCES
        cambio = REDONDEAR(monto_pagado - total_a_pagar, 2 decimales)
    SINO
        cambio = NULO

    DEVOLVER cambio
````

````python
def calcular_cambio(total_a_pagar, monto_pagado):
    if monto_pagado >= total_a_pagar:
        cambio = round(monto_pagado - total_a_pagar, 2)
    else:
        cambio = None
    return cambio
````

````python
# === Tests visibles · Ejercicio 5 ===
assert calcular_cambio(113050.0, 150000) == 36950.0
assert calcular_cambio(113050.0, 100000) is None
print("✅ Ejercicio 5: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 5 ===
assert calcular_cambio(50000, 50000) == 0.0
assert calcular_cambio(99999.99, 100000) == 0.01
print("✅ Ejercicio 5: tests adicionales superados.")
````

## Ejercicio 6 · Simulador completo de caja registradora

Implementa `caja_registradora(precio_unitario, cantidad, metodo_pago, monto_pagado)` que integra todo el flujo y devuelve un **diccionario** con claves:
`subtotal`, `descuento`, `base_gravable`, `iva`, `total`, `cambio`.

Usa `total_compra` y `calcular_cambio`. Si `total_compra` devuelve `None` o el pago es insuficiente, el campo `cambio` es `None`.

**Ejemplo:**
```
caja_registradora(25000, 4, 'efectivo', 150000) ->
{
  'subtotal': 100000, 'descuento': 5000.0,
  'base_gravable': 95000.0, 'iva': 18050.0,
  'total': 113050.0, 'cambio': 36950.0
}
```

````python
FUNCIÓN caja_registradora(precio_unitario, cantidad, metodo_pago, monto_pagado):

    totales = total_compra(precio_unitario, cantidad, metodo_pago)

    SI totales ES NULO ENTONCES
        DEVOLVER diccionario con:
            subtotal: NULO, descuento: NULO, base_gravable: NULO,
            iva: NULO, total: NULO, cambio: NULO

    desempacar totales en: subtotal, descuento, base_gravable, iva, total
    cambio = calcular_cambio(total, monto_pagado)

    DEVOLVER diccionario con:
        subtotal, descuento, base_gravable, iva, total, cambio

````

````python
def caja_registradora(precio_unitario, cantidad, metodo_pago, monto_pagado):
    resultado = total_compra(precio_unitario, cantidad, metodo_pago)

    if resultado is None:
        return {
            'subtotal': None,
            'descuento': None,
            'base_gravable': None,
            'iva': None,
            'total': None,
            'cambio': None
        }

    subtotal, descuento, base_gravable, iva, total = resultado
    cambio = calcular_cambio(total, monto_pagado)

    return {
        'subtotal': subtotal,
        'descuento': descuento,
        'base_gravable': base_gravable,
        'iva': iva,
        'total': total,
        'cambio': cambio
    }
````

````python
# === Tests visibles · Ejercicio 6 ===
_ticket = caja_registradora(25000, 4, 'efectivo', 150000)
assert _ticket['total'] == 113050.0
assert _ticket['cambio'] == 36950.0
print("✅ Ejercicio 6: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 6 ===
_ticket2 = caja_registradora(50000, 2, 'tarjeta', 110000)
assert _ticket2['subtotal'] == 100000
assert _ticket2['cambio'] is None
_ticket3 = caja_registradora(0, 2, 'efectivo', 50000)
assert _ticket3['total'] is None
print("✅ Ejercicio 6: tests adicionales superados.")
````
